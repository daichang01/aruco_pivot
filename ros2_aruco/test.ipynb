```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Initialize the dictionary and parameters
aruco_dict = cv2.aruco.getPredefinedDictionary(cv2.aruco.DICT_5X5_250)
marker_size = 78  # 2 cm in pixels, assuming 100 pixels per cm
marker_spacing = 8  # 2 mm spacing in pixels

# Set up the grid size
num_markers_x = 3  # Number of markers along the x-axis
num_markers_y = 2  # Number of markers along the y-axis
width = num_markers_x * (marker_size + marker_spacing) - marker_spacing
height = num_markers_y * (marker_size + marker_spacing) - marker_spacing

# Create a white image to place the markers on
grid_image = 255 * np.ones((height, width), dtype=np.uint8)

# IDs of the markers from 10 to 15
ids = [10, 11, 12, 13, 14, 15]

# Place each marker in the grid
idx = 0
for y in range(num_markers_y):
    for x in range(num_markers_x):
        marker_id = ids[idx]
        marker_image = cv2.aruco.drawMarker(aruco_dict, marker_id, marker_size)
        x_position = x * (marker_size + marker_spacing)
        y_position = y * (marker_size + marker_spacing)
        grid_image[y_position:y_position + marker_size, x_position:x_position + marker_size] = marker_image
        idx += 1

# Convert the image to RGB (for displaying with matplotlib, which expects this format)
grid_image = cv2.cvtColor(grid_image, cv2.COLOR_GRAY2RGB)

# Show the grid using matplotlib
plt.figure(figsize=(6, 4))
plt.imshow(grid_image)
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Initialize the dictionary and parameters
aruco_dict = cv2.aruco.getPredefinedDictionary(cv2.aruco.DICT_5X5_250)
marker_size = 200  # Increase marker size to 200 pixels for higher resolution
marker_spacing = 488  # Increase spacing to 5 cm in pixels, assuming 100 pixels per cm
border_size = 16  # 2 mm border in pixels

# Set up the grid size including the border
num_markers_x = 2  # Number of markers along the x-axis
num_markers_y = 2  # Number of markers along the y-axis
width = num_markers_x * marker_size + (num_markers_x - 1) * marker_spacing + 2 * border_size
height = num_markers_y * marker_size + (num_markers_y - 1) * marker_spacing + 2 * border_size

# Create a white image to place the markers on, including the border
grid_image = 255 * np.ones((height, width), dtype=np.uint8)

# IDs of the markers from 1 to 4
ids = [1, 2, 3, 4]

# Place each marker in the grid
idx = 0
for y in range(num_markers_y):
    for x in range(num_markers_x):
        marker_id = ids[idx]
        marker_image = cv2.aruco.drawMarker(aruco_dict, marker_id, marker_size)
        x_position = x * (marker_size + marker_spacing) + border_size
        y_position = y * (marker_size + marker_spacing) + border_size
        grid_image[y_position:y_position + marker_size, x_position:x_position + marker_size] = marker_image
        idx += 1

# Add a circle at the center of the image with doubled radius
center_x = width // 2
center_y = height // 2
cv2.circle(grid_image, (center_x, center_y), 4, (0, 0, 0), -1)  # Draw a black circle with a diameter of 8 pixels

# Convert the image to RGB (for displaying with matplotlib, which expects this format)
grid_image_rgb = cv2.cvtColor(grid_image, cv2.COLOR_GRAY2RGB)

# Save the image
cv2.imwrite('/home/daichang/Desktop/teeth_ws/src/aruco_pivot/aruco_images/aruco_board.png', grid_image_rgb)

# Show the grid using matplotlib
plt.figure(figsize=(10, 10))
plt.imshow(grid_image_rgb)
plt.axis('off')
plt.show()
```
